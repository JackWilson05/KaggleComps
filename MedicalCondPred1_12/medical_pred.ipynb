# Clean data
- resolve missing values
- make sure converted to numeric (one hot or ordinal)

```python
# for any missing in age, bmi, bp, glucose -> mean
for col in ["age", "bmi", "blood_pressure", "glucose_levels"]:
    overall_df[col] = overall_df[col].fillna(overall_df[col].mean())
```

```python
# convert non numeric cols

# one hot def + calls
def convert_one_hot(df, colname):
    # create that many cols (and name them og colname_{class})
    # 0 in all except correct one

    df = pd.concat([df, pd.get_dummies(df[colname], prefix=colname)], axis=1).drop(colname, axis=1)
    # get dummies does this for us and then we are dropping the extra col


    return df


# calls -> gender
overall_df = convert_one_hot(overall_df, "gender")



# ordinal def + calls
def convert_ordinal(df, colname, mapping):
    # mappign tells us what the order is

    df[colname] = df[colname].map(mapping)
    # apply on col in vecorized way?
    return df

# calls -> smoking status
smoking_mapping = {"Non-Smoker": 0,
 "Former-Smoker": 1,
 "Smoker": 2}

overall_df = convert_ordinal(overall_df, "smoking_status", smoking_mapping)
```
